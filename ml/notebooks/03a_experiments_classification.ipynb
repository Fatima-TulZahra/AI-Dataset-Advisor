```python
RES = f"{BASE}/results/clf_experiments.csv"
done = set(pd.read_csv(RES)["did"]) if os.path.exists(RES) else set()

for r in lst[lst.task == "classification"].itertuples():
    if r.did in done:
        continue                                  # pehle ho chuka, chhor do
    try:
        df = pd.read_csv(f"{BASE}/data/clf/{r.did}.csv")
        y = df.pop("__target__"); X = df
        feats = extract_meta_features(X, y, "classification")
        scores = {name: run_cv(model, X, y) for name, model in ALGOS_CLF.items()}
        row = {"did": r.did, "split": r.split, **feats, **scores}
        pd.DataFrame([row]).to_csv(RES, mode="a", header=not os.path.exists(RES), index=False)
    except Exception as e:
        print("failed", r.did, e)
```
